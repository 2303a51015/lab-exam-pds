```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("/content/crop_pests_data")
IMAGE_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

def prepare_split_df(split_name):
    image_dir = DATA_DIR / split_name / "images"
    records = []
    for p in image_dir.iterdir():
        if p.is_file() and p.suffix.lower() in IMAGE_EXTS:
            # Filenames look like 'wasp-803-_jpg.rf...' -> class name is the first part
            class_name = p.name.split("-")[0]
            records.append({
                "filepath": str(p),
                "class_name": class_name
            })
    return pd.DataFrame(records)

# Prepare DataFrames directly from the pre-split directories
train_df = prepare_split_df("train")
val_df = prepare_split_df("valid")
test_df = prepare_split_df("test")

# Map class names to unique integer IDs consistently across all splits
all_classes = sorted(train_df["class_name"].unique())
class_to_id = {name: i for i, name in enumerate(all_classes)}

for df in [train_df, val_df, test_df]:
    if not df.empty:
        df["label"] = df["class_name"].map(class_to_id)

print("Classes detected:", all_classes)
print("Training images:", len(train_df))
print("Validation images:", len(val_df))
print("Test images:", len(test_df))
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import DenseNet121
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
import matplotlib.pyplot as plt

# 1. Recreate the correct generators specifying all 13 classes explicitly
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

# 'all_classes' contains the 13 unique classes detected in the dataset
print("Consistent classes to map:", all_classes)

train_datagen = ImageDataGenerator(
    rescale=1.0/255.0,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(rescale=1.0/255.0)

train_generator_df = train_datagen.flow_from_dataframe(
    dataframe=train_df,
    x_col='filepath',
    y_col='class_name',
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    classes=all_classes,  # Ensures 13 outputs
    shuffle=True
)

val_generator_df = val_datagen.flow_from_dataframe(
    dataframe=val_df,
    x_col='filepath',
    y_col='class_name',
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    classes=all_classes,  # Forces 13 outputs even if some are missing in validation
    shuffle=False
)

# 2. Build the DenseNet121 Model with 13 classes
base_model = DenseNet121(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
base_model.trainable = False

x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(128, activation='relu')(x)
x = Dropout(0.2)(x)
predictions = Dense(13, activation='softmax')(x)

model = Model(inputs=base_model.input, outputs=predictions)

# 3. Compile the model
model.compile(
    optimizer=Adam(learning_rate=1e-4),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("Starting model training...")
# 4. Train the model
try:
    history = model.fit(
        train_generator_df,
        validation_data=val_generator_df,
        epochs=3,
        verbose=1
    )

    # 5. Plot the performance curves
    acc = history.history['accuracy']
    val_acc = history.history['val_accuracy']
    loss = history.history['loss']
    val_loss = history.history['val_loss']
    epochs_range = range(len(acc))

    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(epochs_range, acc, label='Train Accuracy', color='blue')
    plt.plot(epochs_range, val_acc, label='Val Accuracy', color='orange')
    plt.title('DenseNet121 - Accuracy Curve')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(epochs_range, loss, label='Train Loss', color='blue')
    plt.plot(epochs_range, val_loss, label='Val Loss', color='orange')
    plt.title('DenseNet121 - Loss Curve')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()
    plt.show()
except Exception as e:
    print("An error occurred during training:", e)
```

```python
import os
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, classification_report, confusion_matrix
)
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint

SEED = 42
IMG_SIZE = (224, 224)
# Set class_names to the detected list of 13 categories from your dataset splits
class_names = all_classes
BATCH_SIZE = 32 if tf.config.list_physical_devices("GPU") else 8

np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))
print("Classes:", class_names)
print("Number of classes:", len(class_names))
print("Batch size:", BATCH_SIZE)
```

```python
import tensorflow as tf

# Extract paths and labels for all splits from our Pandas DataFrames
train_paths = train_df['filepath'].values
train_labels = train_df['label'].values

val_paths = val_df['filepath'].values
val_labels = val_df['label'].values

test_paths = test_df['filepath'].values
test_labels = test_df['label'].values

AUTOTUNE = tf.data.AUTOTUNE

def make_densenet_dataset(paths, labels, training=False):
    ds = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    if training:
        ds = ds.shuffle(
            len(paths), seed=SEED,
            reshuffle_each_iteration=True
        )

    def load_image(path, label):
        image = tf.io.read_file(path)
        image = tf.io.decode_image(
            image, channels=3, expand_animations=False
        )
        image.set_shape([None, None, 3])
        image = tf.image.resize(image, IMG_SIZE)
        image = tf.cast(image, tf.float32)
        return image, label

    ds = ds.map(load_image, num_parallel_calls=AUTOTUNE)
    ds = ds.batch(BATCH_SIZE)
    return ds.prefetch(AUTOTUNE)

train_ds = make_densenet_dataset(
    train_paths, train_labels, training=True
)
val_ds = make_densenet_dataset(val_paths, val_labels)
test_ds = make_densenet_dataset(test_paths, test_labels)

print("Training images:", len(train_paths))
print("Validation images:", len(val_paths))
print("Testing images:", len(test_paths))
```

```python
num_classes = len(class_names)

inputs = layers.Input(shape=(224, 224, 3))

x = layers.RandomFlip("horizontal")(inputs)
x = layers.RandomRotation(0.08)(x)
x = layers.RandomZoom(0.10)(x)
x = layers.RandomContrast(0.10)(x)

x = layers.Lambda(
    tf.keras.applications.densenet.preprocess_input
)(x)

backbone = tf.keras.applications.DenseNet121(
    include_top=False,
    weights="imagenet",
    input_shape=(224, 224, 3)
)
backbone.trainable = False

x = backbone(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.35)(x)
x = layers.Dense(256, activation="relu")(x)
x = layers.Dropout(0.25)(x)

outputs = layers.Dense(
    num_classes, activation="softmax"
)(x)

model = models.Model(inputs, outputs, name="DenseNet121")

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

# 1. Gather all true labels and predictions from the test dataset
y_true = []
y_pred_probs = []

for images, labels in test_ds:
    preds = model.predict(images, verbose=0)
    y_true.extend(labels.numpy())
    y_pred_probs.extend(preds)

y_true = np.array(y_true)
y_pred_probs = np.array(y_pred_probs)
y_pred = np.argmax(y_pred_probs, axis=1)

# 2. Compute evaluation metrics
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, average='macro', zero_division=0)
recall = recall_score(y_true, y_pred, average='macro', zero_division=0)
f1 = f1_score(y_true, y_pred, average='macro', zero_division=0)

try:
    # If class targets are one-hot encoded, calculate multiclass ROC-AUC
    roc_auc = roc_auc_score(tf.keras.utils.to_categorical(y_true, num_classes=13), y_pred_probs, multi_class='ovr')
except Exception:
    roc_auc = 0.0

# 3. Define metrics to display
metric_names = [
    "Test Accuracy",
    "Macro Precision",
    "Macro Recall",
    "Macro F1-score",
    "Macro ROC-AUC"
]

metric_values = [
    accuracy, precision, recall, f1, roc_auc
]

# 4. Generate the performance bar plot
plt.figure(figsize=(10, 5))
bars = plt.bar(metric_names, metric_values, color=['#4285F4', '#EA4335', '#FBBC05', '#34A853', '#9C27B0'])

plt.ylim(0, 1.1)
plt.ylabel("Score")
plt.title("DenseNet121 - Test Performance Metrics")
plt.xticks(rotation=25, ha="right")
plt.grid(axis="y", alpha=0.3)

for bar, value in zip(bars, metric_values):
    if np.isfinite(value):
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            value + 0.015,
            f"{value:.3f}",
            ha="center"
        )

plt.tight_layout()
plt.savefig("/content/densenet121_metrics_bar.png", dpi=300)
plt.show()
```
